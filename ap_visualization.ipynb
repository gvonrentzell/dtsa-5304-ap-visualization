```python
import pandas as pd
import altair as alt

alt.renderers.enable("jupyter", offline=True)
```

```python
ap_data = pd.read_csv("data/processed/ap_state_data.csv")

ap_data.head()
```

```python
state_select = alt.selection_point(
    fields=["state"],
    toggle=True
)

scatter = alt.Chart(ap_data).mark_circle(size=100).encode(
    x=alt.X(
        "participation_2025:Q",
        title="AP Participation Rate"
    ),
    y=alt.Y(
        "pass_rate_2025:Q",
        title="AP Exam 3+ Rate"
    ),
    tooltip=[
        "state:N",
        "participation_2025:Q",
        "pass_rate_2025:Q"
    ],
    color=alt.condition(
        state_select,
        alt.value("steelblue"),
        alt.value("lightgray")
    ),
).add_params(state_select)

scatter
```

```python
participation_bar = (
    alt.Chart(ap_data)
    .transform_filter(state_select)
    .mark_bar()
    .encode(
        x=alt.X(
            "participation_2025:Q",
            title="AP Participation Rate",
            scale=alt.Scale(domain=[0, 0.40]),
            axis=alt.Axis(format=".0%")
        ),
        y=alt.Y(
            "state:N",
            title=None
        ),
        tooltip=[
            "state:N",
            alt.Tooltip(
                "participation_2025:Q",
                title="Participation",
                format=".1%"
            )
        ]
    )
    .properties(
        width=350,
        height=180,
        title="AP Participation"
    )
)

pass_rate_bar = (
    alt.Chart(ap_data)
    .transform_filter(state_select)
    .mark_bar()
    .encode(
        x=alt.X(
            "pass_rate_2025:Q",
            title="AP Exam Pass Rate",
            scale=alt.Scale(domain=[0, 0.90]),
            axis=alt.Axis(format=".0%")
        ),
        y=alt.Y(
            "state:N",
            title=None
        ),
        tooltip=[
            "state:N",
            alt.Tooltip(
                "pass_rate_2025:Q",
                title="AP Exam Pass Rate",
                format=".1%"
            )
        ]
    )
    .properties(
        width=350,
        height=180,
        title="AP Exam Pass Rate"
    )
)

state_select = alt.selection_point(
    fields=["state"],
    toggle="true",
    empty=False
)

scatter = alt.Chart(ap_data).mark_circle(size=100).encode(
    x=alt.X(
        "participation_2025:Q",
        title="AP Participation Rate"
    ),
    y=alt.Y(
        "pass_rate_2025:Q",
        title="AP Exam Pass Rate"
    ),
    tooltip=[
        "state:N",
        "participation_2025:Q",
        "pass_rate_2025:Q"
    ],
    color=alt.condition(
        state_select,
        alt.value("steelblue"),
        alt.value("lightgray")
    ),
).add_params(state_select)

dashboard = (
    scatter | (participation_bar & pass_rate_bar)
    ).properties(
    title={
        "text": "Exploring AP Participation and Performance by State",
        "subtitle": [
            "Click a state to select it. Shift-click additional states to compare.",
            "For the clearest comparison, select up to 10 states."
        ]
    }
)

dashboard
```
