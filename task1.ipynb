```python
import pandas as pd
import altair as alt

ap_data = pd.read_csv("data/processed/ap_state_data.csv")

ap_data.head()
```

# Exploring AP Participation and Performance by State

Explore how AP participation and exam performance vary across states.

### Explore the data:
- Click a state to select it.
- **Shift-click** to select multiple states.
- Compare the selected states using the charts below.
- Hover over a state for additional information.

```python
# --------------------------------------------------
# TASK 1: PARTICIPATION AND PERFORMANCE
# --------------------------------------------------

# Create the selection FIRST so every chart below can use it
state_select = alt.selection_point(
    fields=["state"],
    toggle="true",
    empty=False
)


# --------------------------------------------------
# SCATTERPLOT
# --------------------------------------------------

scatter = (
    alt.Chart(ap_data)
    .mark_circle(size=100)
    .encode(
        x=alt.X(
            "participation_2025:Q",
            title="AP Participation Rate"
        ),
        y=alt.Y(
            "pass_rate_2025:Q",
            title="AP Exam Pass Rate"
        ),
        tooltip=[
            "state:N",
            alt.Tooltip(
                "participation_2025:Q",
                title="Participation",
                format=".1%"
            ),
            alt.Tooltip(
                "pass_rate_2025:Q",
                title="AP Exam Pass Rate",
                format=".1%"
            )
        ],
        color=alt.condition(
            state_select,
            alt.value("steelblue"),
            alt.value("lightgray")
        )
    )
    .add_params(state_select)
)


# --------------------------------------------------
# PARTICIPATION BAR CHART
# --------------------------------------------------

participation_bar = (
    alt.Chart(ap_data)
    .transform_filter(state_select)
    .mark_bar()
    .encode(
        x=alt.X(
            "participation_2025:Q",
            title="AP Participation Rate",
            scale=alt.Scale(domain=[0, 0.40]),
            axis=alt.Axis(format=".0%")
        ),
        y=alt.Y(
            "state:N",
            title=None
        ),
        tooltip=[
            "state:N",
            alt.Tooltip(
                "participation_2025:Q",
                title="Participation",
                format=".1%"
            )
        ]
    )
    .properties(
        width=350,
        height=180,
        title="AP Participation"
    )
)


# --------------------------------------------------
# PASS RATE BAR CHART
# --------------------------------------------------

pass_rate_bar = (
    alt.Chart(ap_data)
    .transform_filter(state_select)
    .mark_bar()
    .encode(
        x=alt.X(
            "pass_rate_2025:Q",
            title="AP Exam Pass Rate",
            scale=alt.Scale(domain=[0, 0.90]),
            axis=alt.Axis(format=".0%")
        ),
        y=alt.Y(
            "state:N",
            title=None
        ),
        tooltip=[
            "state:N",
            alt.Tooltip(
                "pass_rate_2025:Q",
                title="AP Exam Pass Rate",
                format=".1%"
            )
        ]
    )
    .properties(
        width=350,
        height=180,
        title="AP Exam Pass Rate"
    )
)


# --------------------------------------------------
# COMBINE DASHBOARD
# --------------------------------------------------

task1_dashboard = (
    scatter
    | (participation_bar & pass_rate_bar)
).properties(
    title={
        "text": "Exploring AP Participation and Performance by State",
        "subtitle": [
            "Click a state to select it. Shift-click additional states to compare.",
            "For the clearest comparison, select up to 10 states."
        ]
    }
)

task1_dashboard
```
