```python
import pandas as pd
import altair as alt

ap_data = pd.read_csv("data/processed/ap_state_data.csv")
```

# How Does the Definition of AP Success Change the Story?

Compare states using multiple measures of AP participation and performance.

### Explore the data:
- **Click** a state to select it.
- **Shift-click** to select multiple states.
- **Drag over a group of states** to compare them.
- Compare the selected states across the four measures below.
- Hover over a state or bar for additional information.

```python
## Task 2: Comparing Measures of AP Success


task2_click = alt.selection_point(
    fields=["state"],
    toggle="true",
    empty=False
)

task2_brush = alt.selection_interval(
    encodings=["x", "y"],
    empty=False
)


task2_base = alt.Chart(ap_data).encode(
    x=alt.X(
        "participation_2025:Q",
        title="AP Participation Rate",
        axis=alt.Axis(format=".0%")
    ),
    y=alt.Y(
        "pass_rate_2025:Q",
        title="AP Exam Pass Rate",
        axis=alt.Axis(format=".0%")
    )
)

task2_points = task2_base.mark_circle(size=175).encode(
    color=alt.condition(
        task2_click | task2_brush,
        alt.value("steelblue"),
        alt.value("lightgray")
    ),
    tooltip=[
        "state:N",
        alt.Tooltip("participation_2025:Q", title="Participation", format=".1%"),
        alt.Tooltip("pass_rate_2025:Q", title="Pass Rate", format=".1%")
    ]
)

task2_labels = task2_base.mark_text(
    fontSize=8
).encode(
    text="state_abbr:N"
)

task2_scatter = (
    (task2_points + task2_labels)
    .add_params(task2_click, task2_brush)
    .properties(width=725, height=420)
)

task2_high_score_bar = (
    alt.Chart(ap_data)
    .transform_filter(task2_click | task2_brush)
    .mark_bar()
    .encode(
        x=alt.X(
            "high_score_rate_2025:Q",
            title="AP Exams Scoring 4–5",
            scale=alt.Scale(domain=[0, 0.40]),
            axis=alt.Axis(format=".0%")
        ),
        y=alt.Y(
            "state:N",
            title=None
        ),
        tooltip=[
            "state:N",
            alt.Tooltip(
                "high_score_rate_2025:Q",
                title="AP Exams Scoring 4–5",
                format=".1%"
            )
        ]
    )
    .properties(
        width=350,
        height=100,
        title="AP Exams Scoring 4–5"
    )
)


task2_pass_rate_bar = (
    alt.Chart(ap_data)
    .transform_filter(task2_click | task2_brush)
    .mark_bar()
    .encode(
        x=alt.X(
            "pass_rate_2025:Q",
            title="AP Exam Pass Rate",
            scale=alt.Scale(domain=[0, 0.90]),
            axis=alt.Axis(format=".0%")
        ),
        y=alt.Y(
            "state:N",
            title=None
        ),
        tooltip=[
            "state:N",
            alt.Tooltip(
                "pass_rate_2025:Q",
                title="AP Exam Pass Rate",
                format=".1%"
            )
        ]
    )
    .properties(
        width=350,
        height=100,
        title="AP Exam Pass Rate"
    )
)

task2_participation_bar = (
    alt.Chart(ap_data)
    .transform_filter(task2_click | task2_brush)
    .mark_bar()
    .encode(
        x=alt.X(
            "participation_2025:Q",
            title="AP Participation Rate",
            scale=alt.Scale(domain=[0, 0.40]),
            axis=alt.Axis(format=".0%")
        ),
        y=alt.Y(
            "state:N",
            title=None
        ),
        tooltip=[
            "state:N",
            alt.Tooltip(
                "participation_2025:Q",
                title="AP Participation Rate",
                format=".1%"
            )
        ]
    )
    .properties(
        width=350,
        height=100,
        title="AP Participation Rate"
    )
)


task2_exams_bar = (
    alt.Chart(ap_data)
    .transform_filter(task2_click | task2_brush)
    .mark_bar()
    .encode(
        x=alt.X(
            "exams_per_examinee_2025:Q",
            title="Exams per AP Examinee"
        ),
        y=alt.Y(
            "state:N",
            title=None
        ),
        tooltip=[
            "state:N",
            alt.Tooltip(
                "exams_per_examinee_2025:Q",
                title="Exams per AP Examinee",
                format=".2f"
            )
        ]
    )
    .properties(
        width=350,
        height=100,
        title="Exams per AP Examinee"
    )
)

task2_dashboard = (
    task2_scatter
    &
    (task2_participation_bar | task2_pass_rate_bar)
    &
    (task2_high_score_bar | task2_exams_bar)
).properties(
    title={
        "text": "How Does the Definition of AP Success Change the Story?",
        "subtitle": [
            "Compare states using multiple measures of AP participation and performance.",
            "Click a state, Shift-click multiple states, or drag over a group of states to compare."
        ],
        "fontSize": 20,
        "subtitleFontSize": 12,
        "anchor": "middle"
    }
)

task2_dashboard
```
