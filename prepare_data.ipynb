```python
performance = pd.read_excel(
    "data/raw/ap-admin-report-2025-performance.xlsx",
    header=None
)

performance.columns = [
    "blank",
    "state_or_group",
    "score",
    "2025",
    "2024",
    "2020",
    "2015"
]

performance.head(15)
```

```python
state_rows = performance[
    performance["state_or_group"].isin(states)
]

state_rows[["state_or_group"]].head(10)
```

```python
state_rows.shape
```

```python

florida_index = state_rows[
    state_rows["state_or_group"] == "Florida"
].index[0]

florida_index
```

```python
performance.loc[florida_index:florida_index + 8]
```

```python
performance_records = []

for state in states:
    state_index = performance[
        performance["state_or_group"] == state
    ].index[0]

    block = performance.loc[state_index + 1:state_index + 6]

    record = {
        "state": state,
        "score5_2025": block.iloc[0]["2025"],
        "score4_2025": block.iloc[1]["2025"],
        "score3_2025": block.iloc[2]["2025"],
        "score2_2025": block.iloc[3]["2025"],
        "score1_2025": block.iloc[4]["2025"],
        "total_exams_2025": block.iloc[5]["2025"]
    }

    performance_records.append(record)
```

```python
performance_states = pd.DataFrame(performance_records)
performance_states.shape
```

```python
performance_states = pd.DataFrame(performance_records)

performance_states.head()
```

```python
performance_states.shape
```

```python
performance_states[
    performance_states["state"] == "Florida"
]
```

```python
performance_states["pass_rate_2025"] = (
    performance_states["score5_2025"]
    + performance_states["score4_2025"]
    + performance_states["score3_2025"]
) / performance_states["total_exams_2025"]
```

```python
performance_states["high_score_rate_2025"] = (
    performance_states["score5_2025"]
    + performance_states["score4_2025"]
) / performance_states["total_exams_2025"]
```

```python
performance_states[
    performance_states["state"] == "Florida"
]
```

```python
ap_data = pd.merge(
    participation_states,
    performance_states,
    on="state"
)

ap_data.head()
```

```python
ap_data.shape
```

```python
ap_data[
    ap_data["state"] == "Florida"
]
```

```python
ap_data["exams_per_examinee_2025"] = (
    ap_data["total_exams_2025"]
    / ap_data["examinees_2025"]
)
```
